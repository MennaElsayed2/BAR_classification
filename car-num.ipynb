```python

# Step 4: Create directories for storing models and data
os.makedirs('models', exist_ok=True)
os.makedirs('data', exist_ok=True)

# Step 5: Load and train the YOLOv8 model
print("Training the model...")
model = YOLO('yolov8n.pt')  # Load pretrained YOLOv8 model

# Train the model
model.train(
    data=os.path.join(dataset.location, 'data.yaml'),
    epochs=50,  # You can adjust this number
    imgsz=640,
    batch=16,
    name='car_number_detector'
)
```

```python

# Step 6: Test the model with an image
def test_model(image_path):

    # Load image
    image = cv2.imread('https://www.google.com/url?sa=i&url=https%3A%2F%2Fen.wikipedia.org%2Fwiki%2FVehicle_registration_plates_of_Egypt&psig=AOvVaw2o3spj2Rb1lOD0_B7clNr1&ust=1744143712797000&source=images&cd=vfe&opi=89978449&ved=0CBQQjRxqFwoTCPDS8uXfxowDFQAAAAAdAAAAABAE')
    if image is None:
        print(f"Error: Could not load image at {'https://www.google.com/url?sa=i&url=https%3A%2F%2Fen.wikipedia.org%2Fwiki%2FVehicle_registration_plates_of_Egypt&psig=AOvVaw2o3spj2Rb1lOD0_B7clNr1&ust=1744143712797000&source=images&cd=vfe&opi=89978449&ved=0CBQQjRxqFwoTCPDS8uXfxowDFQAAAAAdAAAAABAE'}")
        return

    # Run detection
    results = model(image)

    # Process and display results
    for result in results:
        boxes = result.boxes
        for box in boxes:
            # Get coordinates
            x1, y1, x2, y2 = box.xyxy[0]
            confidence = box.conf[0]

            # Draw rectangle
            cv2.rectangle(image, (int(x1), int(y1)), (int(x2), int(y2)), (0, 255, 0), 2)

            # Draw confidence score
            cv2.putText(image, f'{confidence:.2f}', (int(x1), int(y1)-10),
                        cv2.FONT_HERSHEY_SIMPLEX, 0.5, (0, 255, 0), 2)

    # Display the result
    cv2_imshow(image)

# Step 7: Example usage
print("\nTesting the model...")
# You can test with an image from your dataset
test_image_path = os.path.join(dataset.location, 'test', 'test_image.jpg')  # Replace with actual test image path
test_model(test_image_path)
```

```python

# Step 8: Real-time detection using webcam (if running locally)
def real_time_detection():

    cap = cv2.VideoCapture(0)

    while True:
        ret, frame = cap.read()
        if not ret:
            break

        # Run detection
        results = model(frame)

        # Process and display results
        for result in results:
            boxes = result.boxes
            for box in boxes:
                x1, y1, x2, y2 = box.xyxy[0]
                confidence = box.conf[0]

                # Draw rectangle
                cv2.rectangle(frame, (int(x1), int(y1)), (int(x2), int(y2)), (0, 255, 0), 2)

                # Draw confidence score
                cv2.putText(frame, f'{confidence:.2f}', (int(x1), int(y1)-10),
                            cv2.FONT_HERSHEY_SIMPLEX, 0.5, (0, 255, 0), 2)

        # Display the frame
        cv2_imshow(frame)

        # Break the loop if 'q' is pressed
        if cv2.waitKey(1) & 0xFF == ord('q'):
            break

    cap.release()
    cv2.destroyAllWindows()
```
